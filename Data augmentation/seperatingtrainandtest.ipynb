```python
import os
import shutil
import re
```

```python

def create_blue_blob_train_images(source_folder, destination_folder):
    """
    Creates a new folder containing images from the source folder where the number after 'BC' in the filename is even.

    Parameters:
        source_folder (str): Path to the folder containing the original blue blob images.
        destination_folder (str): Path to the folder where selected images will be copied.

    Returns:
        None
    """
    # Ensure the destination folder exists
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
        print(f"Created destination folder: {destination_folder}")

    # Get a list of all files in the source folder
    all_files = os.listdir(source_folder)

    # Filter files where the number after 'BC' is even
    selected_files = []
    for file_name in all_files:
        match = re.search(r'BC(\d+)', file_name)  # Look for 'BC' followed by a number
        if match:
            number = int(match.group(1))  # Extract the number
            if number % 2 == 0:  # Check if the number is even
                selected_files.append(file_name)

    # Copy selected files to the destination folder
    for file_name in selected_files:
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(destination_folder, file_name)
        shutil.copy(source_path, destination_path)

    print(f"Copied {len(selected_files)} files to {destination_folder}.")

# Example usage
source_folder = "Augmented_images"  # Replace with the actual path to your source folder
destination_folder = "blue_blob_train_images_new"  # Replace with the desired destination folder name

create_blue_blob_train_images(source_folder, destination_folder)
```

```python
def create_remaining_blue_blob_images(source_folder, excluded_folder, destination_folder):
    """
    Creates a new folder containing images from the source folder that are NOT in the excluded folder.

    Parameters:
        source_folder (str): Path to the folder containing the original blue blob images.
        excluded_folder (str): Path to the folder containing the excluded images.
        destination_folder (str): Path to the folder where remaining images will be copied.

    Returns:
        None
    """
    # Ensure the destination folder exists
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
        print(f"Created destination folder: {destination_folder}")

    # Get a list of all files in the source folder
    all_files = set(os.listdir(source_folder))

    # Get a list of all files in the excluded folder
    excluded_files = set(os.listdir(excluded_folder))

    # Find the remaining files
    remaining_files = all_files - excluded_files

    # Copy remaining files to the destination folder
    for file_name in remaining_files:
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(destination_folder, file_name)
        shutil.copy(source_path, destination_path)

    print(f"Copied {len(remaining_files)} files to {destination_folder}.")

# Example usage
source_folder = "Augmented_images"  # Replace with the actual path to your source folder
excluded_folder = "blue_blob_train_images"  # Replace with the path to the excluded folder
destination_folder = "blue_blob_test_images"  # Replace with the desired destination folder name

create_remaining_blue_blob_images(source_folder, excluded_folder, destination_folder)
```

```python
def create_galaxy_train_and_test_folders(source_folder, train_folder, test_folder):
    """
    Creates two folders: one for training (every other pair of files) and one for testing (remaining pairs).

    Parameters:
        source_folder (str): Path to the folder containing the original galaxy images.
        train_folder (str): Path to the folder where training images will be copied.
        test_folder (str): Path to the folder where testing images will be copied.

    Returns:
        None
    """
    # Ensure the train and test folders exist
    if not os.path.exists(train_folder):
        os.makedirs(train_folder)
        print(f"Created training folder: {train_folder}")
    if not os.path.exists(test_folder):
        os.makedirs(test_folder)
        print(f"Created testing folder: {test_folder}")

    # Get a sorted list of all files in the source folder
    all_files = sorted(os.listdir(source_folder))
    print(len(all_files)*.8)
    # Group files into pairs
    file_pairs = [all_files[i:i+2] for i in range(0, len(all_files), 2)]
    split = int(len(all_files)*.8)

    # Split pairs into train (every other pair) and test (remaining pairs)
    train_pairs = file_pairs[:split]  # Every other pair
    test_pairs = file_pairs[split:]  # Remaining pairs

    # Flatten the pairs into a single list of files
    train_files = [file for pair in train_pairs for file in pair]
    test_files = [file for pair in test_pairs for file in pair]

    # Copy train files to the train folder
    for file_name in train_files:
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(train_folder, file_name)
        shutil.copy(source_path, destination_path)

    # Copy test files to the test folder
    for file_name in test_files:
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(test_folder, file_name)
        shutil.copy(source_path, destination_path)

    print(f"Copied {len(train_files)} files to {train_folder}.")
    print(f"Copied {len(test_files)} files to {test_folder}.")

# Example usage
source_folder = "trial_galaxies_image_new"  # Replace with the actual path to your source folder
train_folder = "galaxy_train"  # Replace with the desired training folder name
test_folder = "galaxy_test"  # Replace with the desired testing folder name

create_galaxy_train_and_test_folders(source_folder, train_folder, test_folder)
```

```python
def keep_one_in_every_20_files(source_folder, destination_folder):
    """
    Copies one file for every 20 files from the source folder to the destination folder.

    Parameters:
        source_folder (str): Path to the folder containing the original files.
        destination_folder (str): Path to the folder where selected files will be copied.

    Returns:
        None
    """
    # Ensure the destination folder exists
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
        print(f"Created destination folder: {destination_folder}")

    # Get a sorted list of all files in the source folder
    all_files = sorted(os.listdir(source_folder))

    # Select one file for every 20 files
    selected_files = all_files[::2]

    # Copy selected files to the destination folder
    for file_name in selected_files:
        source_path = os.path.join(source_folder, file_name)
        destination_path = os.path.join(destination_folder, file_name)
        shutil.copy(source_path, destination_path)

    print(f"Copied {len(selected_files)} files to {destination_folder}.")

# Example usage
source_folder = "galaxy_test"  # Replace with the actual path to your source folder
destination_folder = "galaxies_test_images_subset"  # Replace with the desired destination folder name

keep_one_in_every_20_files(source_folder, destination_folder)
```

```python
import os
import shutil

def delete_matching_files(source_folder, reference_folder, destination_folder):
    """
    Deletes files in the source folder that also exist in the reference folder,
    and copies the remaining files to the destination folder.

    Parameters:
        source_folder (str): Path to the folder where files will be checked.
        reference_folder (str): Path to the folder containing reference files.
        destination_folder (str): Path to the folder where non-matching files will be copied.

    Returns:
        None
    """
    # Ensure the destination folder exists
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
        print(f"Created destination folder: {destination_folder}")

    # Get a set of filenames in the reference folder
    reference_files = set(os.listdir(reference_folder))

    # Iterate through files in the source folder
    for file_name in os.listdir(source_folder):
        source_path = os.path.join(source_folder, file_name)
        if file_name in reference_files:
            print(f"Deleted: {file_name}")
            #os.remove(source_path)  # Uncomment this line to actually delete the file
        else:
            destination_path = os.path.join(destination_folder, file_name)
            shutil.copy(source_path, destination_path)
            print(f"Kept: {file_name}")

# Example usage
source_folder = "image_arnd_bb_augmented"  # Replace with the path to your source folder
reference_folder = "galaxies_test_images_subset"  # Replace with the path to your reference folder
destination_folder = "galaxy_train"  # Replace with the path to your destination folder

delete_matching_files(source_folder, reference_folder, destination_folder)
```
