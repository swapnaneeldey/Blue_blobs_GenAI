```python
import os
from PIL import Image  # Import the Image module from Pillow
```

```python

# --- Configuration ---

# 1. Set the path to the folder containing your ORIGINAL images
source_folder = 'stat_rotating_background' 

destination_folder = 'Downsample_stat_rot_images' 

# 3. Set the target size (width, height) in pixels
target_size = (256, 256)

# 4. List of image file extensions to process
image_extensions = "jpeg"


# --- Process Images ---
print(f"\nStarting downsampling process...")
print(f"Source folder:      {source_folder}")
print(f"Destination folder: {destination_folder}")
print(f"Target size:        {target_size}")
print("-" * 30)

processed_count = 0
error_count = 0

# Get a list of all files in the source directory
try:
    filenames = os.listdir(source_folder)
except FileNotFoundError:
    print(f"Error: Source folder not found at {source_folder}")
    exit()
except Exception as e:
    print(f"Error accessing source folder {source_folder}: {e}")
    exit()

# Loop through each file
for filename in filenames:
    # Check if the file has one of the allowed image extensions (case-insensitive)
    if filename.lower().endswith(image_extensions):
        source_path = os.path.join(source_folder, filename)
        dest_path = os.path.join(destination_folder, filename) # Keep the same filename

        # Double-check if it's actually a file (and not a subdirectory)
        if os.path.isfile(source_path):
            try:
                # Open the image using Pillow
                with Image.open(source_path) as img:
                    # --- Handle potential transparency (RGBA) ---
                    # If saving as JPG, need to convert RGBA/P modes to RGB first
                    save_format = dest_path.split('.')[-1].lower()
                    if save_format in ['jpg', 'jpeg'] and img.mode not in ['RGB']:
                        print(f"  Converting {filename} from mode '{img.mode}' to 'RGB' for JPG saving.")
                        img = img.convert('RGB')

                    # Resize the image - Use a high-quality downsampling filter
                    # Image.Resampling.LANCZOS or Image.Resampling.BICUBIC are good choices
                    # (Note: Older Pillow versions might use Image.LANCZOS or Image.BICUBIC)
                    try:
                        # Try the modern Resampling enum (Pillow >= 9.1.0)
                        resampling_filter = Image.Resampling.LANCZOS
                        resized_img = img.resize(target_size, resample=resampling_filter)
                    except AttributeError:
                        # Fallback for older Pillow versions
                        print("Note: Using older Pillow resampling filter name (e.g., Image.LANCZOS)")
                        resampling_filter = Image.LANCZOS
                        resized_img = img.resize(target_size, resample=resampling_filter)

                    # Save the resized image to the destination folder
                    resized_img.save(dest_path)
                    print(f"  Successfully processed: {filename} -> {dest_path}")
                    processed_count += 1

            except Exception as e:
                print(f"  Error processing {filename}: {e}")
                error_count += 1
        else:
            # This handles cases where an item in the directory isn't a file
            pass # Silently skip directories or other non-file items

print("-" * 30)
print("Downsampling complete!")
print(f"Successfully processed: {processed_count} images.")
if error_count > 0:
    print(f"Failed to process:    {error_count} images (check errors above).")
print(f"Downsampled images saved in: {destination_folder}")
```
